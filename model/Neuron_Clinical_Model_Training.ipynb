### Abstract
Over 10 million individuals globally live with neurodegenerative motor disorders such as Parkinson's Disease and motor ataxia. Early diagnosis is severely hindered by the high cost and inaccessibility of specialized clinical gait and tremor labs. 

**NEURON** addresses this through millisecond keystroke dynamics. When humans type on standard physical or virtual keyboards, involuntary micro-hesitations, finger contact hold times, and inter-key latency flight times reflect underlying neuromuscular control. This notebook trains a **4-layer Deep Multi-Layer Perceptron (ANN)** on 14 clinical biometric features extracted from keystroke time-series data.
